# `OllamaLLMProvider`

```python
from rich.pretty import pprint

from aymurai.llm_providers import OllamaLLMProvider
```

```python
MODEL_NAME = "llama3.2:3b"
SYSTEM_PROMPT = "Eres un asistente legal especializado en anonimización y desambiguación de entidades."

provider = OllamaLLMProvider(model=MODEL_NAME, system_prompt=SYSTEM_PROMPT)
provider
```

## Chunking

```python
sample_text = (
    "La jueza fijó audiencia para la próxima semana y solicitó a las partes presentar "
    "documentación adicional relacionada con la causa. El objetivo es determinar si existieron "
    "violaciones al protocolo vigente y establecer medidas de protección para las víctimas."
)

chunks = provider.chunk_text(sample_text, max_tokens=10, overlap=0)
for chunk in chunks:
    print(f"Chunk {chunk.index}: {chunk.token_count} tokens")
    pprint(chunk)
```

## Generation

```python
prompt = (
    "Resume en español el siguiente extracto de un expediente judicial y menciona los actores involucrados:"
    + sample_text
)

try:
    summary_response = provider.generate(prompt)
    print(summary_response.text)
    pprint(summary_response.metadata)

except Exception as exc:
    print("No se pudo contactar a Ollama:", exc)
```

## Streaming

```python
try:
    stream_prompt = "Explica en detalle el procedimiento legal descrito."
    pieces = []
    for event in provider.stream(stream_prompt):
        print(event.text, end="", flush=True)
        pieces.append(event.text)
    print()
    full_text = "".join(pieces)

except Exception as exc:
    print("La transmisión no está disponible:", exc)
```
