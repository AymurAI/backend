```python
import locale
import platform

locales_a_probar = ['es_AR.UTF-8', 'es_ES.UTF-8', 'es_MX.UTF-8', 'spanish']

for loc in locales_a_probar:
    try:
        locale.setlocale(locale.LC_ALL, loc)
        break
    except locale.Error:
        continue

print(f"Sistema: {platform.system()}")
print(f"Locale actual: {locale.getlocale()}")
print(f"Default Locale: {locale.getdefaultlocale()}")
```

```python
# 1. Forzamos a Python a que crea que "es_AR.UTF-8" es lo mismo que el locale básico
# Esto intercepta la llamada que falla en patterns.py:4
def mock_setlocale(category, value=None):
    return "C" # 'C' es el locale universal que siempre existe

locale.setlocale = mock_setlocale

# 2. Opcional: Si quieres ser más preciso, intenta usar el español genérico
try:
    # En Mac a veces es 'es_ES.UTF-8' o solo 'es_AR' (sin .UTF-8)
    # Aquí intentamos forzar uno que funcione
    import _locale
    _locale._setlocale(locale.LC_ALL, 'en_US.UTF-8') 
except:
    pass

print("✅ Hack de locale aplicado. Ya puedes importar aymurai.")
```
