```python
import sys
from pathlib import Path

from aymurai.evaluation.metrics import evaluate_prediction_directories

project_root = Path.cwd().parents[2]
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
```

```python
# Ajusta estas rutas antes de ejecutar
TEST_DIR = Path("/resources/data/restricted/disambiguation-eval/test")
PREDS_DIR = Path(
    "/resources/data/restricted/disambiguation-eval/preds/preds_phi4:14b_pv1_251125_1716"
)
```

```python
# Normalizando los textos
results, average = evaluate_prediction_directories(TEST_DIR, PREDS_DIR)

if not results:
    print("No se encontraron pares test/pred que evaluar.")
else:
    for doc_id, score, metrics in results:
        print(f"Documento: {doc_id}")
        print(f"  Score global: {score:.4f}")
        for metric_name, value in metrics.items():
            print(f"  {metric_name}: {value:.4f}")
        print()

    print(f"Promedio Score_global: {average:.4f}")
```

```python
# Sin Normalizar los textos
results, average = evaluate_prediction_directories(TEST_DIR, PREDS_DIR, normalize=False)

if not results:
    print("No se encontraron pares test/pred que evaluar.")
else:
    for doc_id, score, metrics in results:
        print(f"Documento: {doc_id}")
        print(f"  Score global: {score:.4f}")
        for metric_name, value in metrics.items():
            print(f"  {metric_name}: {value:.4f}")
        print()

    print(f"Promedio Score_global: {average:.4f}")
```
